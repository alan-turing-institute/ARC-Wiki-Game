```python
from __future__ import annotations

import matplotlib.pyplot as plt
import pandas as pd
```

```python
embed_wiki_df = pd.read_csv(
    "../result/table/embed_model_wiki_summary_stats.csv", sep="\t"
)
prompt_wiki_df = pd.read_csv("../result/table/prompt_wiki_summary_stats.csv", sep="\t")
tfidf_wiki_df = pd.read_csv("../result/table/tfidf_wiki_summary_stats.csv", sep="\t")
```

```python
embed_wiki_df["BarColor"] = "skyblue"
prompt_wiki_df["BarColor"] = "lightgreen"
tfidf_wiki_df["BarColor"] = "lightcoral"
```

```python
wiki_df = pd.concat([prompt_wiki_df, tfidf_wiki_df, embed_wiki_df], axis=0)

wiki_df
```

```python
en_wiki_df = wiki_df[wiki_df["Language"] == "English"]
```

```python
def plot_bar_chart(df, metric, title, y_label, bar_label="Model", figsize=(7, 4)):
    fig = plt.figure(figsize=figsize)
    ax = fig.add_subplot(111)
    ax.bar(
        df[bar_label],
        df[metric],
        yerr=df[metric + "_std"],
        capsize=5,
        color=df["BarColor"],
        zorder=3,
    )
    ax.set_ylabel(y_label)
    ax.set_title(title)

    ax.set_xticklabels(df[bar_label], rotation=30, ha="right")

    ax.grid()

    plt.tight_layout()

    plt.show()
```

```python
plot_bar_chart(
    en_wiki_df.sort_values("Mean"),
    "Mean",
    "Step Count for English Wikipedia",
    "Mean Step Count",
)
```

```python
plot_bar_chart(
    en_wiki_df.sort_values("Success_rate", ascending=False),
    "Success_rate",
    "Success Rate for English Wikipedia",
    "Success Rate",
)
```

```python
selected_df = wiki_df[
    wiki_df["Model"].isin(
        ["ChatGPT 4o", "Intfloat Multilingual-E5-Large", "TFIDF wiki-20171001"]
    )
]
selected_df["Model_Lang"] = selected_df["Model"] + " (" + selected_df["Language"] + ")"
selected_df.head(2)
```

```python
plot_bar_chart(
    selected_df.sort_values("Success_rate"),
    "Success_rate",
    "Success Rate for Different Languages",
    "Success Rate",
    bar_label="Model_Lang",
    figsize=(4, 4),
)
```

```python
plot_bar_chart(
    selected_df.sort_values("Mean"),
    "Mean",
    "Step Count for Different Languages",
    "Mean Step Count",
    bar_label="Model_Lang",
    figsize=(4, 4),
)
```

```python
fig = plt.figure(figsize=(7, 4))
ax = fig.add_subplot(111)

ax.plot([-0.5, len(wiki_df) + 0.5], [1, 1], ":k", zorder=2)

ax.errorbar(
    range(len(wiki_df)), wiki_df["n"], wiki_df["n_std"], fmt=".", zorder=3, capsize=2
)

ax.set_ylim(0, 2)
ax.set_xlim(-0.5, len(wiki_df) + 0.5)

ax.set_ylabel("Required Correct Steps")
ax.set_xlabel("Models")

ax.set_title("Step parameter for Negative Binomial Distribution")

plt.show()
```

```python
fig = plt.figure(figsize=(7, 4))
ax = fig.add_subplot(111)


ax.errorbar(
    wiki_df["Mean"], wiki_df["p"], wiki_df["p_std"], fmt=".", zorder=3, capsize=2
)


ax.set_ylim(0.0, 0.05)
ax.set_xlim(28, 44)

ax.set_ylabel("Success Probability")
ax.set_xlabel("Step Count")

ax.set_title("Probability Parameter for Negative Binomial Distribution")

plt.show()
```
