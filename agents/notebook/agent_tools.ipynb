# Agent Tools Demenstration

```python
from __future__ import annotations

import os

from agents.agent_models.tools import AgentTools

main_dir = os.path.dirname(os.getcwd())
hdf5_file_path = os.path.join(main_dir, "training_data/wiki_forward.hdf5")
matidx_info_path = os.path.join(main_dir, "training_data/matidx_sum_table.csv")
summary_path = os.path.join(main_dir, "training_data/summary.csv")
```
